
🧩 4단계: 리스트 안의 딕셔너리 안의 리스트

구조: [ { "key": [ , , ] } , { "key": [ , , ] } ]
 → 리스트 안에 딕셔너리가 있고, 그 딕셔너리의 값이 리스트인 형태

```python
#문제 1 예제 1의 학교 정보를 schools 변수에 만들고, 첫 번째 학교의 이름을 출력하세요.
schools =  [ { "name": "서울고등학교",
              "classes": { "1학년": ["김민수", "이영희"],
                           "2학년": ["박철수", "최지영"] }
                },
             { "name": "부산고등학교",
              "classes": { "1학년": ["정다은", "강민호"],
                          "2학년": ["윤서준"] } }
             ]
```

```python
schools[0]["name"]
```

```python
# 문제 2 예제 2의 회사 정보를 companies 변수에 만들고, 첫 번째 회사("A Corp")의 "Sales" 부서 직원 목록을 출력하세요.
companies = [
  {
    "company": "A Corp",
    "departments": {
      "Sales": ["Alice", "Eve"],
      "IT": ["Bob", "David"]
    }
  },
  {
    "company": "B Corp",
    "departments": {
      "HR": ["Charlie"],
      "Marketing": ["Diana", "Frank"]
    }
  }
]
```

```python
companies[0]["departments"]["Sales"]
```

```python
# 문제 3 예제 3의 쇼핑몰 정보를 malls 변수에 만들고, 두 번째 쇼핑몰("전자몰")의 "주변기기" 카테고리의 첫 번째 상품을 출력하세요.
malls =  [
    { "mall": "패션몰",
      "categories":
       { "상의": ["티셔츠", "셔츠", "니트"],
         "하의": ["청바지", "슬랙스"] }
      },
    { "mall": "전자몰",
      "categories":
      { "컴퓨터": ["노트북", "데스크탑"],
        "주변기기": ["마우스", "키보드", "모니터"] }
     }
    ]
```

```python
malls[1]["categories"]["주변기기"][0]
```

```python
# 문제 4 예제 4의 도서관 정보를 libraries 변수에 만들고, 두 번째 도서관의 "경제" 장르에서 두 번째 책을 출력하세요.
libraries =  [
     { "name": "중앙도서관",
      "genres": { "소설": ["해리포터", "반지의제왕"],
                 "과학": ["코스모스", "이기적유전자"] } },
     { "name": "시립도서관",
      "genres": { "역사": ["총균쇠", "사피엔스"],
                 "경제": ["국부론", "자본론"] } }
     ]
```

```python
libraries[1]["genres"]["경제"][1]
```

```python
# 문제 5 예제 5의 식당 정보를 restaurants 변수에 만들고, 첫 번째 식당("한식당")의 "구이류" 메뉴 중 세 번째 메뉴를 출력하세요.
restaurants = [
     { "name": "한식당",
      "menu": { "찌개류": ["김치찌개", "된장찌개"],
               "구이류": ["불고기", "삼겹살", "갈비"] } },
     { "name": "중식당",
      "menu": { "면류": ["짜장면", "짬뽕"],
               "요리": ["탕수육", "깐풍기"] } }
    ]
```

```python
restaurants[0]["menu"]["구이류"][2]
```


📌 학습 정리

2단계 인덱싱 패턴
리스트[인덱스]["키"]
예: students[0]["name"]

3단계 인덱싱 패턴
딕셔너리["키"][인덱스]
예: classes["1반"][0]

4단계 인덱싱 패턴
리스트[인덱스]["키"]["키"][인덱스]
예: schools[0]["classes"]["1학년"][0]
